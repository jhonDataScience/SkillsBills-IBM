Separa el dataset en dos variables, 'x' e 'y'.

```python
# Carga el dataset "diamonds" de seaborn
diamantes = sns.load_dataset("diamonds")
diamantes.head()

# Resuelve aqui el ejercicio 3.1

diamantes = sns.load_dataset("diamonds")

feature_cols = ['depth', 'table', 'price', 'x', 'y', 'z']

target_col = 'carat'

# Separación del dataset:
X = diamantes[feature_cols]
y = diamantes[target_col]

print("--- Primeros 5 registros de X ---")
print(X.head())
print("\n--- Primeros 5 registros de y ---")
print(y.head())
```

#### Ejercicio 3.2
Separa el dataset provisto en conjunto de entrenamiento y de test, dando un 40% de los datos al conjunto de test

```python
# Resuelve aqui el ejercicio 3.2

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.4,
    random_state=42
)

# Opcional: Mostrar las dimensiones para confirmar la división
print("--- Dimensiones de los conjuntos ---")
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

#### Ejercicio 3.3
Crea un modelo de regresión lineal y entrenalo con los conjuntos de datos separados en el ejercicio anterior. La variable objetivo corresponde a la columna 'carat'

```python
# Resuelve aqui el ejercicio 3.3

model = LinearRegression()

print("Entrenando el modelo...")
model.fit(X_train, y_train)
print("¡Entrenamiento completado!")
print("\n--- Coeficientes del modelo (Pendientes) ---")
feature_names = X_train.columns
coefficients = model.coef_
for name, coef in zip(feature_names, coefficients):
    print(f"{name}: {coef:.4f}")

print(f"Intercepto (Bias): {model.intercept_:.4f}")
```

#### Ejercicio 3.4

#### Ejercicio 6.3
Separa el dataset provisto en conjunto de entrenamiento y de test, dando un 40% de los datos al conjunto de test.

```python
# Resuelve aquí el ejercicio 6.3

df = pd.read_csv(url, header=None)
df.columns = ["Sex", "Length", "Diameter", "Height", "Whole weight", "Shucked weight", "Viscera weight", "Shell weight", "Rings",]
X = df.drop('Sex', axis=1)
y = df['Sex'] # Variable objetivo categórica


le = LabelEncoder()
y_encoded = le.fit_transform(y)


X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.4,
    random_state=42,
    stratify=y_encoded
)

print("--- Dimensiones de los conjuntos ---")
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

#### Ejercicio 6.4
Crea un modelo de regresión logística y entrenalo con los conjuntos de datos separados en el ejercicio anterior. Utiliza el valor 3 para el parámetro 'n_neighbors'

```python
# Resuelve aquí el ejercicio 6.4

warnings.filterwarnings('ignore')

knn_model = KNeighborsClassifier(n_neighbors=3)

print("Entrenando el modelo K-Nearest Neighbors (KNN) con k=3...")
knn_model.fit(X_train, y_train)
print("¡Entrenamiento completado!")
```

#### Ejercicio 6.5
Muestra la matriz de confusión resultante para la salida del ejercicio anterior

```python
# Resuelve aquí el ejercicio 6.5

try:
    y_pred = knn_model.predict(X_test)
except NameError:
    y_pred = knn_model.predict(X_test)

class_names = np.unique(y)

cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=class_names
)


plt.figure(figsize=(8, 8))
disp.plot(cmap=plt.cm.Blues, values_format='d')
plt.title('Matriz de Confusión para el Modelo KNN (k=3)')
plt.show()


print("\n--- Matriz de Confusión (Array Numérico) ---")
print(cm)
```
